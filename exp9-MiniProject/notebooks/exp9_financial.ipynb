```python
import pandas as pd
import numpy as np

# Load original dataset
df = pd.read_csv("../financial/Customer_Financial_Profiles.csv")

print("Original Shape:", df.shape)


# --------------------------------------------------
# 1. STANDARDIZE COLUMN NAMES
# --------------------------------------------------

df.columns = [
    "CustomerID",
    "CurrentAge",
    "BirthYear",
    "BirthMonth",
    "Gender",
    "Address",
    "PerCapitaIncome",
    "YearlyIncome",
    "TotalDebt",
    "CreditScore",
    "NumCreditCards",
    "TransactionID",
    "Date",
    "ClientID",
    "CardID",
    "Amount",
    "UseChip",
    "MerchantID",
    "MerchantCity",
    "MerchantState",
    "Zip"
]


# --------------------------------------------------
# 2. DATA TYPE CONVERSION
# --------------------------------------------------

numeric_cols = [
    "CustomerID",
    "CurrentAge",
    "BirthYear",
    "BirthMonth",
    "PerCapitaIncome",
    "YearlyIncome",
    "TotalDebt",
    "CreditScore",
    "NumCreditCards",
    "Amount"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")


# Convert date to datetime
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")


# IDs are identifiers, not measures
id_cols = [
    "TransactionID",
    "ClientID",
    "CardID",
    "MerchantID"
]

for col in id_cols:
    df[col] = df[col].astype(str).str.strip()


# Clean categorical columns
categorical_cols = [
    "Gender",
    "UseChip",
    "MerchantCity",
    "MerchantState"
]

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()


# Zip code is a geographic identifier
df["Zip"] = df["Zip"].astype(str).str.replace(".0", "", regex=False)


# --------------------------------------------------
# 3. REMOVE EXACT DUPLICATES
# --------------------------------------------------

duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicates before cleaning:", duplicates_before)
print("Duplicates after cleaning:", duplicates_after)


# --------------------------------------------------
# 4. FEATURE ENGINEERING
# --------------------------------------------------

# Debt-to-Income Ratio
df["DebtToIncomeRatio"] = np.where(
    df["YearlyIncome"] > 0,
    (df["TotalDebt"] / df["YearlyIncome"]) * 100,
    0
)


# Age Groups
df["AgeGroup"] = pd.cut(
    df["CurrentAge"],
    bins=[18, 25, 35, 45, 55, 65, 100],
    labels=[
        "19-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "66+"
    ]
)


# Credit Score Category
df["CreditScoreCategory"] = pd.cut(
    df["CreditScore"],
    bins=[0, 579, 669, 739, 799, 900],
    labels=[
        "Poor",
        "Fair",
        "Good",
        "Very Good",
        "Excellent"
    ]
)


# Income Category
df["IncomeCategory"] = pd.qcut(
    df["YearlyIncome"],
    q=4,
    labels=[
        "Low Income",
        "Lower-Middle Income",
        "Upper-Middle Income",
        "High Income"
    ]
)


# Debt Status
df["DebtStatus"] = np.where(
    df["TotalDebt"] == 0,
    "No Debt",
    "Has Debt"
)


# Transaction Amount Category
df["TransactionAmountCategory"] = pd.cut(
    df["Amount"],
    bins=[0, 1000, 5000, 10000, np.inf],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)


# Date-based features
df["TransactionYear"] = df["Date"].dt.year
df["TransactionMonth"] = df["Date"].dt.month
df["TransactionMonthName"] = df["Date"].dt.month_name()
df["TransactionQuarter"] = "Q" + df["Date"].dt.quarter.astype(str)


# --------------------------------------------------
# 5. VALIDATION
# --------------------------------------------------

print("\n=== CLEANED DATASET ===")
print("Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nCredit Score Range:")
print(df["CreditScore"].min(), "to", df["CreditScore"].max())

print("\nAge Range:")
print(df["CurrentAge"].min(), "to", df["CurrentAge"].max())

print("\nTransaction Amount Range:")
print(df["Amount"].min(), "to", df["Amount"].max())

print("\nDebt-to-Income Ratio Range:")
print(
    round(df["DebtToIncomeRatio"].min(), 2),
    "to",
    round(df["DebtToIncomeRatio"].max(), 2)
)

print("\nFinal Columns:")
print(df.columns.tolist())

display(df.head())
```

```python
# Detailed validation of engineered financial features

print("=== AGE GROUP ===")
print(df["AgeGroup"].value_counts().sort_index())

print("\n=== CREDIT SCORE CATEGORY ===")
print(df["CreditScoreCategory"].value_counts())

print("\n=== INCOME CATEGORY ===")
print(df["IncomeCategory"].value_counts())

print("\n=== DEBT STATUS ===")
print(df["DebtStatus"].value_counts())

print("\n=== TRANSACTION AMOUNT CATEGORY ===")
print(df["TransactionAmountCategory"].value_counts())

print("\n=== TRANSACTION YEAR ===")
print(df["TransactionYear"].value_counts().sort_index())

print("\n=== PAYMENT METHOD ===")
print(df["UseChip"].value_counts())

print("\n=== MERCHANT STATE ===")
print(df["MerchantState"].value_counts())

print("\n=== FINANCIAL SUMMARY ===")
display(
    df[
        [
            "PerCapitaIncome",
            "YearlyIncome",
            "TotalDebt",
            "CreditScore",
            "NumCreditCards",
            "Amount",
            "DebtToIncomeRatio"
        ]
    ].describe().T
)
```

```python
# Save cleaned financial dataset

output_path = "../financial/Customer_Financial_Profiles_Cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File:", output_path)
print("Shape:", df.shape)
```

```python
import pandas as pd
import mysql.connector

# Load cleaned dataset
df = pd.read_csv("../financial/Customer_Financial_Profiles_Cleaned.csv")

# Convert date to proper MySQL date format
df["Date"] = pd.to_datetime(df["Date"]).dt.date

print("Rows to insert:", len(df))
print("Columns:", len(df.columns))
```

```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root",
    database="financial_analytics"
)

cursor = conn.cursor()

print("MySQL connection successful!")
```

```python
insert_sql = """
INSERT INTO customer_financial_profiles (
    CustomerID,
    CurrentAge,
    BirthYear,
    BirthMonth,
    Gender,
    Address,
    PerCapitaIncome,
    YearlyIncome,
    TotalDebt,
    CreditScore,
    NumCreditCards,
    TransactionID,
    Date,
    ClientID,
    CardID,
    Amount,
    UseChip,
    MerchantID,
    MerchantCity,
    MerchantState,
    Zip,
    DebtToIncomeRatio,
    AgeGroup,
    CreditScoreCategory,
    IncomeCategory,
    DebtStatus,
    TransactionAmountCategory,
    TransactionYear,
    TransactionMonth,
    TransactionMonthName,
    TransactionQuarter
)
VALUES (
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s
)
"""

records = [
    tuple(row)
    for row in df[
        [
            "CustomerID",
            "CurrentAge",
            "BirthYear",
            "BirthMonth",
            "Gender",
            "Address",
            "PerCapitaIncome",
            "YearlyIncome",
            "TotalDebt",
            "CreditScore",
            "NumCreditCards",
            "TransactionID",
            "Date",
            "ClientID",
            "CardID",
            "Amount",
            "UseChip",
            "MerchantID",
            "MerchantCity",
            "MerchantState",
            "Zip",
            "DebtToIncomeRatio",
            "AgeGroup",
            "CreditScoreCategory",
            "IncomeCategory",
            "DebtStatus",
            "TransactionAmountCategory",
            "TransactionYear",
            "TransactionMonth",
            "TransactionMonthName",
            "TransactionQuarter"
        ]
    ].itertuples(index=False, name=None)
]
```

```python
for i in range(0, len(records), 5000):
    batch = records[i:i+5000]

    cursor.executemany(insert_sql, batch)
    conn.commit()

    print(f"Inserted {min(i + 5000, len(records))} / {len(records)} rows")
```
