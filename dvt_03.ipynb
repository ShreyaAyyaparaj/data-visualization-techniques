```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/Disaster_dataset.csv", sep=";")

print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
numeric_columns = [
    "Total Events",
    "Affected",
    "Deaths",
    "DamageOriginal",
    "DamageAdjusted",
    "CPI"
]

for col in numeric_columns:
    numeric_columns = [
        "Total Events",
        "Total Affected",
        "Total Deaths",
        "Total Damage (USD, original)",
        "Total Damage (USD, adjusted)",
        "CPI"
    ]

    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
print(df.columns.tolist())
```

```python
df.rename(columns={
    "Disaster Type": "DisasterType",
    "Disaster Subtype": "DisasterSubtype",
    "Total Affected": "Affected",
    "Total Deaths": "Deaths",
    "Total Damage (USD, original)": "DamageOriginal",
    "Total Damage (USD, adjusted)": "DamageAdjusted"
}, inplace=True)
```

```python
numeric_columns = [
    "Year",
    "Total Events",
    "Affected",
    "Deaths",
    "DamageOriginal",
    "DamageAdjusted",
    "CPI"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df["Affected"] = df["Affected"].fillna(0)
df["Deaths"] = df["Deaths"].fillna(0)
df["DamageOriginal"] = df["DamageOriginal"].fillna(0)
df["DamageAdjusted"] = df["DamageAdjusted"].fillna(0)

df["DisasterSubtype"] = df["DisasterSubtype"].fillna("Unknown")

df["CPI"] = df["CPI"].fillna(df["CPI"].median())
```

```python
print(df.isnull().sum())
```

```python
damage = (
    df.groupby("Country")["DamageOriginal"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(damage)
```

```python
damage = (
    df.groupby("Country")["DamageOriginal"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(damage)
```
