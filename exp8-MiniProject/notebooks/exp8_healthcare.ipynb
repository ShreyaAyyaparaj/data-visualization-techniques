```python
df.rename(columns={
    "Patient Number": "PatientID",
    "Heart Rate (bpm)": "HeartRate",
    "SpO2 Level (%)": "SpO2",
    "Systolic Blood Pressure (mmHg)": "SystolicBP",
    "Diastolic Blood Pressure (mmHg)": "DiastolicBP",
    "Body Temperature (°C)": "BodyTemperature",
    "Fall Detection": "FallDetection",
    "Predicted Disease": "PredictedDisease",
    "Data Accuracy (%)": "DataAccuracy",
    "Heart Rate Alert": "HeartRateAlert",
    "SpO2 Level Alert": "SpO2Alert",
    "Blood Pressure Alert": "BloodPressureAlert",
    "Temperature Alert": "TemperatureAlert"
}, inplace=True)

print(df.columns.tolist())
```

```python
numeric_columns = [
    "PatientID",
    "HeartRate",
    "SpO2",
    "SystolicBP",
    "DiastolicBP",
    "BodyTemperature",
    "DataAccuracy"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_columns].dtypes)
```

```python
print("Missing values after numeric conversion:")
print(df[numeric_columns].isnull().sum())
```

```python
print(
    "Total missing values after conversion:",
    df.isnull().sum().sum()
)
```

```python
categorical_columns = [
    "FallDetection",
    "PredictedDisease",
    "HeartRateAlert",
    "SpO2Alert",
    "BloodPressureAlert",
    "TemperatureAlert"
]

for col in categorical_columns:
    print("\n", col)
    print(df[col].value_counts())
```

```python
for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
print("Heart Rate range:")
print(df["HeartRate"].min(), "to", df["HeartRate"].max())

print("\nSpO2 range:")
print(df["SpO2"].min(), "to", df["SpO2"].max())

print("\nSystolic BP range:")
print(df["SystolicBP"].min(), "to", df["SystolicBP"].max())

print("\nDiastolic BP range:")
print(df["DiastolicBP"].min(), "to", df["DiastolicBP"].max())

print("\nBody Temperature range:")
print(df["BodyTemperature"].min(), "to", df["BodyTemperature"].max())

print("\nData Accuracy range:")
print(df["DataAccuracy"].min(), "to", df["DataAccuracy"].max())
```

```python
numeric_analysis_columns = [
    "HeartRate",
    "SpO2",
    "SystolicBP",
    "DiastolicBP",
    "BodyTemperature",
    "DataAccuracy"
]

for col in numeric_analysis_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} potential statistical outliers")
```

```python
df["PulsePressure"] = (
    df["SystolicBP"] - df["DiastolicBP"]
)
```

```python
df["MeanArterialPressure"] = (
    df["SystolicBP"] +
    (2 * df["DiastolicBP"])
) / 3
```

```python
alert_columns = [
    "HeartRateAlert",
    "SpO2Alert",
    "BloodPressureAlert",
    "TemperatureAlert"
]

df["AbnormalVitalCount"] = (
    df[alert_columns]
    .eq("ABNORMAL")
    .sum(axis=1)
)
```

```python
df["HealthAlertLevel"] = pd.cut(
    df["AbnormalVitalCount"],
    bins=[-1, 0, 1, 2, 4],
    labels=[
        "Normal",
        "Low",
        "Moderate",
        "High"
    ]
)
```

```python
df["AccuracyCategory"] = pd.cut(
    df["DataAccuracy"],
    bins=[84, 89, 94, 97, 100],
    labels=[
        "Acceptable",
        "Good",
        "Very Good",
        "Excellent"
    ]
)
```

```python
import mysql.connector

print("MySQL connector imported successfully!")
```

```python
cleaned_df = pd.read_csv(
    "../healthcare/Patient_Data_Healthcare_Cleaned.csv"
)

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
```

```python
from getpass import getpass

mysql_password = getpass("Enter MySQL root password: ")

connection = mysql.connector.connect(
    host="localhost",
    port=3306,
    user="root",
    password=mysql_password,
    database="healthcare_analytics"
)

print("MySQL connection successful!")
```

```python
cursor = connection.cursor()

cursor.execute("""
    SELECT COUNT(*)
    FROM patient_monitoring
""")

current_count = cursor.fetchone()[0]

print("Current records in MySQL:", current_count)
```

```python
insert_query = """
INSERT INTO patient_monitoring (
    PatientID,
    HeartRate,
    SpO2,
    SystolicBP,
    DiastolicBP,
    BodyTemperature,
    FallDetection,
    PredictedDisease,
    DataAccuracy,
    HeartRateAlert,
    SpO2Alert,
    BloodPressureAlert,
    TemperatureAlert,
    PulsePressure,
    MeanArterialPressure,
    AbnormalVitalCount,
    HealthAlertLevel,
    AccuracyCategory
)
VALUES (
    %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s
)
"""

batch_size = 5000
total_rows = len(cleaned_df)

for start in range(0, total_rows, batch_size):
    end = min(start + batch_size, total_rows)

    batch = cleaned_df.iloc[start:end]
    records = [
        tuple(row)
        for row in batch.itertuples(index=False, name=None)
    ]

    cursor.executemany(insert_query, records)
    connection.commit()

    print(f"Inserted {end} / {total_rows} records")

print("Data import completed successfully!")
```
